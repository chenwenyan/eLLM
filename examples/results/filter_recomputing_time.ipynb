```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['ps.fonttype'] = 42	
import numpy as np
import pandas as pd
import warnings
import seaborn as sns
import re, csv, torch
warnings.filterwarnings("ignore")

linestyles = ['-', '--', '-.', ':']
markers = ['o', 's', 'D', '^', 'v', 'x', '+', '*', 'p', 'h']
hatches = ['////', '\\\\\\', '||||', '----', '++++', 'xx', 'oo', 'OO', '.', '**']
```

```python
# 从log文件内提取数据
# recomputing time: 1.9875839948654175 ms, tokens: 20

def extract_data(log_file):
    with open(log_file, 'r') as f:
        lines = f.readlines()
    recomputing_times = []
    tokens = []
    for line in lines:
        if re.match(r'recomputing time: \d+\.\d+ ms, tokens: \d+', line):
            recomputing_time = float(re.findall(r'recomputing time: (\d+\.\d+) ms', line)[0])
            token = int(re.findall(r'tokens: (\d+)', line)[0])
            recomputing_times.append(recomputing_time)
            tokens.append(token)
    # 追加写入csv文件
    with open('recomputing_time.csv', 'a') as f:
        writer = csv.writer(f)
        for i in range(len(tokens)):
            writer.writerow([tokens[i], recomputing_times[i]])
    # df = pd.DataFrame({ 'token_num': tokens, 'recomputing_time': recomputing_times})
    df = pd.read_csv('recomputing_time.csv')
    df = df.sort_values(by='token_num')
    df = df.groupby('token_num').mean().reset_index()
    df.to_csv('recomputing_time.csv', index=False)
```

```python
log_file = '../server.log'
extract_data(log_file)
```
