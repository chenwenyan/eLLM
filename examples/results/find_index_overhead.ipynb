```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['ps.fonttype'] = 42	
import numpy as np
import pandas as pd
import warnings
import seaborn as sns
import re, csv, torch
warnings.filterwarnings("ignore")

linestyles = ['-', '--', '-.', ':']
markers = ['o', 's', 'D', '^', 'v', 'x', '+', '*', 'p', 'h']
hatches = ['////', '\\\\\\', '||||', '----', '++++', 'xx', 'oo', 'OO', '.', '**']
```

```python

def parse_time_to_csv(file_path, output_path):
    with open(file_path, 'r') as f:
        lines = f.readlines()
    allocated_block_times = []
    forward_times = []
    for line in lines:
        if line.startswith('allocate block time is '):
            allocated_time = float(line.split('allocate block time is ')[1].split(' ')[0])
            allocated_block_times.append(allocated_time)
        if line.startswith('forward time: '):
            forward_time = float(line.split('forward time: ')[1].split(' ')[0])
            forward_times.append(forward_time)
    
    if allocated_block_times and forward_times:
        print('find_block_index_time mean: ', float('%.2f' % np.mean(allocated_block_times)), 'max: ', float('%.2f' % np.max(allocated_block_times)), 'min: ', float('%.2f' % np.min(allocated_block_times)))
        print('forward_time mean: ', float('%.2f' % np.mean(forward_times)), 'max: ', float('%.2f' % np.max(forward_times)), 'min: ', float('%.2f' % np.min(forward_times)))
        print('ratio: ', float('%.6f' % ((np.mean(allocated_block_times) * 80 )/ np.mean(forward_times) * 100)), '%')
    else:
        print('No valid data found in the log file.')

# file_path = './log_overhead/llama2-13b-server-reqrate100.log'
file_path = '../server.log'
output_path = 'time.csv'
parse_time_to_csv(file_path, output_path)
```
