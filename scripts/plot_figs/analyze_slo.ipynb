```python
def extract_and_count(log_filename, threshold):
    tpots_list = []
    
    pattern = r'tpots: \[(.*?)\]'
    
    with open(log_filename, 'r', encoding='utf-8') as file:
        for line in file:
            match = re.search(pattern, line)
            if match:
                list_str = match.group(1)
                list_str = list_str.replace(' ', '')
                tpots_list = list(map(float, list_str.split(',')))
                break 
    
    count = sum(1 for num in tpots_list if num < threshold)
    
    return count
```

```python
system_names = ['vllm-recompute', 'vllm-swap', 'hcache', 'ellm']
system_labels = ['Recompute', 'Swap', 'HCache', 'eLLM']
model_names = ['meta-llama_Llama-2-13b-chat-hf', 'meta-llama_Llama-2-70b-chat-hf']
# num_promts = 3024
num_promts = [3024, 184]
data_names = ['sharegpt','paper_assistant']
# now = datetime.datetime.now()
current_date = '2025-05-12'

model_labels = ['Llama-2-13B', 'Llama-2-70B']

for i, system_name in enumerate(system_names):
    for j, model_name in enumerate(model_names):
        for k, data_name in enumerate(data_names):
            log_path = f'{DATASET_PATH}/{system_name}/{data_name}/{model_name}_client_{num_promts[k]}_recompute_1gpu.log'
            count = extract_and_count(log_path, 0.06)
            print(f"{system_labels[i]} {model_labels[j]} {data_names[k]}: {count} times")
        
```
